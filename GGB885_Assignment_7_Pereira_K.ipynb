```python
#Updating to include information on trends among the highest paid athletes in the world
url = "https://raw.githubusercontent.com/ali-ce/datasets/master/Most-paid-athletes/Athletes.csv"

athletes = pd.read_csv(url)

athletes.head()
```

```python
# Stripping "$" and ","
athletes["Total Pay"] = athletes["Total Pay"].str.replace("$", "", regex=False).str.replace(",", "", regex=False).astype(float)
```

```python
# What is the correlation between an athlete's height and their total pay?
athletes[["Height (cm)", "Total Pay"]].corr()
```

```python
#Creating an age column for cleaner readability
athletes["Age"] = 2026 - athletes["Year of birth"]
```

```python
# What is the difference between the 75th percentile and 25th percentile of top-paid athlete ages?
athletes["Age"].quantile(0.75) - athletes["Age"].quantile(0.25)
```

```python
# Compare the mean total pay by sport. What sport has the highest paid athletes in this data, on average?
athletes.groupby("Sport")["Total Pay"].mean().sort_values(ascending=False).head(1)
```

```python
# Updating notebook. The Basketball Account Team has identified an API that will return detailed player statistics by season.
import requests

url = "https://api.server.nbaapi.com/api/playertotals?season=2024&page=1&pageSize=1000"

response = requests.get(url)

data = response.json()

data
```

```python
nba_players = pd.DataFrame(data["data"])

nba_players.head()
```

```python
#What is the maximum number of games started by a player in the 2024 season?
nba_players["gamesStarted"].max()
```

```python
#How many players started that number of games in the 2024 season?
(nba_players["gamesStarted"] == nba_players["gamesStarted"].max()).sum()
```

```python
all_players = []

for page in range(1, 20):
    url = f"https://api.server.nbaapi.com/api/playertotals?season=2024&page={page}&pageSize=1000"
    response = requests.get(url)
    page_data = response.json()

    all_players.extend(page_data["data"])

nba_players = pd.DataFrame(all_players)

nba_players.shape
```

```python
# Create a list of the 10 underused players with the highest "stocks per game."
nba_players["stocks_per_game"] = (
    (nba_players["steals"] + nba_players["blocks"])
    / nba_players["games"]
)

nba_players["minutes_per_game"] = (
    nba_players["minutesPg"]
    / nba_players["games"]
)

top10 = (
    nba_players[nba_players["minutes_per_game"] < 25]
    .sort_values("stocks_per_game", ascending=False)
    .head(10)
)

top10[["playerName", "stocks_per_game", "minutes_per_game"]]
```
